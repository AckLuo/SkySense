```python
import sys
sys.path.insert(0, '../src')

import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

from m1_signal import sine_wave, square_wave, fft_spectrum, demonstrate_aliasing
print('SkySense M1 环境就绪')
```

## 实验 1：正弦信号的频谱

改变 `freq` 观察频谱峰值位置的变化。

```python
freq = 20      # 试试改成 50、100
fs = 1000
t, x = sine_wave(freq, fs, 1.0)

f, mag, _ = fft_spectrum(x, fs)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(t[:200], x[:200]); axes[0].set_title(f'时域 {freq} Hz')
axes[1].plot(f, mag); axes[1].set_title('频域'); axes[1].set_xlim(0, 200)
plt.show()
```

## 实验 2：亲手制造混叠

当采样率 < 2×信号频率时，发生混叠。试试把 `fs_sample` 调到 40（30Hz 信号）。

```python
result = demonstrate_aliasing(freq=30, fs=1000, fs_sample=40, duration=0.5)
print(f"原始频率: {result['freq']} Hz")
print(f"采样率: {result['fs_sample']} Hz, Nyquist: {result['nyquist']} Hz")
print(f"混叠后表观频率: {result['alias_freq']} Hz")

plt.figure(figsize=(10, 3))
plt.plot(result['t'], result['x'], alpha=0.5, label='原始 30Hz')
plt.plot(result['t_sampled'], result['x_sampled'], 'ro-', label='采样')
plt.legend(); plt.title('混叠现象'); plt.show()
```
